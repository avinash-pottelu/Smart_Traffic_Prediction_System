```python
columns = [
    'id',
    'date',
    'time',
    'city',
    'area',
    'vehicle_count',
    'avg_speed',
    'congestion_level',
    'weather',
    'accident_count',
    'road_type',
    'signal_wait_time',
    'travel_time'
]

df = pd.read_csv("C:/Users/dell/OneDrive/TrafficVision_Analytics/dataset/traffic_data.csv", names=columns)
```

```python
df.head()
```

```python
city_traffic = df.groupby('city')['vehicle_count'].sum()

plt.figure(figsize=(10,5))
city_traffic.plot(kind='bar')

plt.title('Total Traffic by City')
plt.xlabel('City')
plt.ylabel('Vehicle Count')

plt.show()
```

```python
congestion = df['congestion_level'].value_counts()

plt.figure(figsize=(6,6))
congestion.plot(kind='pie', autopct='%1.1f%%')

plt.title('Traffic Congestion Distribution')

plt.ylabel('')

plt.show()
```

```python
df['hour'] = pd.to_datetime(df['time'], format='%H:%M:%S').dt.hour

hourly_traffic = df.groupby('hour')['vehicle_count'].sum()

plt.figure(figsize=(10,5))
hourly_traffic.plot(kind='line', marker='o')

plt.title('Peak Traffic Hours')
plt.xlabel('Hour of Day')
plt.ylabel('Total Vehicle Count')

plt.grid(True)

plt.show()
```

```python
weather_traffic = df.groupby('weather')['vehicle_count'].mean()

plt.figure(figsize=(8,5))
weather_traffic.plot(kind='bar')

plt.title('Weather Impact on Traffic')
plt.xlabel('Weather')
plt.ylabel('Average Vehicle Count')

plt.show()
```

```python
accidents = df.groupby('city')['accident_count'].sum()

plt.figure(figsize=(10,5))
accidents.plot(kind='bar')

plt.title('Accident Hotspots by City')
plt.xlabel('City')
plt.ylabel('Total Accidents')

plt.show()
```

```python
avg_speed = df.groupby('city')['avg_speed'].mean()

plt.figure(figsize=(10,5))
avg_speed.plot(kind='bar')

plt.title('Average Speed by City')
plt.xlabel('City')
plt.ylabel('Average Speed')

plt.show()
```

```python
top_areas = df.groupby('area')['vehicle_count'].sum().sort_values(ascending=False).head(5)

plt.figure(figsize=(10,5))
top_areas.plot(kind='bar')

plt.title('Top 5 Busiest Areas')
plt.xlabel('Area')
plt.ylabel('Vehicle Count')

plt.show()
```
